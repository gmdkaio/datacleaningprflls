```python
import pandas as pd
import sys
sys.path.append('..')  
df = pd.read_csv('../data/df_maquinas.csv')
df_vendedores = pd.read_csv('../data/df_vendedores.csv')
```

```python
print(df.columns)
```

Limpar vendedores


```python
from dicts.dictconfigs import name_mapping, id_replacer

# Renomear coluna se necessário
if 'Consultor ' in df.columns:
    df = df.rename(columns={'Consultor ': 'Vendedor'})

# Padronizar nomes dos vendedores e adicionar coluna com ID do vendedor
df['Vendedor'] = df['Vendedor'].replace(name_mapping)
df['vendedor_id'] = df['Vendedor'].map(id_replacer)
```

```python
# Verificar diferenças entre colunas originais e reais
valor_diff = df[df['Valor'] != df['Valor Real']]
qtd_diff = df[df['Qtd'] != df['Qtd Real']]

if len(valor_diff) > 0:
    print("Diferenças em Valor:")
    display(valor_diff[['Proposta Comercial', 'Valor', 'Valor Real', 'Vendedor', 'Local']])

if len(qtd_diff) > 0:
    print("Diferenças em Quantidade:")
    display(qtd_diff[['Proposta Comercial', 'Qtd', 'Qtd Real', 'Vendedor', 'Local']])

if len(valor_diff) == 0 and len(qtd_diff) == 0:
    print("Nenhuma diferença encontrada")
```

```python
# Remover colunas desnecessárias
df = df.drop(['Valor Real', 'Qtd Real', 'Produção', 'Nº'], axis=1)
```

```python
# Renomear status Cliente para Aberto
df['Status'] = df['Status'].replace('Cliente', 'Aberto')
# Limpar CNPJ: manter apenas números e garantir que seja string
df['CNPJ'] = df['CNPJ'].astype(str).str.replace(r'[^0-9]', '', regex=True)
# Corrigir formato da coluna Per: substituir ' por /
df['Per'] = df['Per'].str.replace("'", "/")
# Renomear colunas
df = df.rename(columns={
    'Qtd': 'Quantidade',
    'Máquinas': 'Maquinas', 
    'Per': 'Data',
    'Proposta Comercial': 'Codigo',
    'Nome do Cliente': 'Cliente'
})
```
